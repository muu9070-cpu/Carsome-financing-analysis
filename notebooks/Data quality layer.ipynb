```python
submission_dates = [
    'Loan Submission Date1',
    'Loan Submission Date2',
    'Loan Submission Date3',
    'Loan Submission Date4',
    'Loan Submission Date5'
]

date_sequence_issues = []

for i in range(len(submission_dates) - 1):
    current = submission_dates[i]
    next_col = submission_dates[i + 1]

    mask = (
        df[current].notna()
        & df[next_col].notna()
        & (df[next_col] < df[current])
    )

    date_sequence_issues.append({
        'Check': f'{next_col} earlier than {current}',
        'Cases': mask.sum()
    })

date_sequence_issues = pd.DataFrame(date_sequence_issues)

date_sequence_issues
```
